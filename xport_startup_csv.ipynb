```python
silver_df = spark.readStream.table(
    "`startup-data`.`startup-raw-data`.silver_startup_funding"
)

query = (
    silver_df.writeStream
    .format("csv")
    .option("header", "true")
    .option(
        "checkpointLocation",
        "s3://startup-data-s3-bucket/checkpoints/silver_csv_export/"
    )
    .outputMode("append")
    .trigger(availableNow=True)
    .start(
        "s3://startup-data-s3-bucket/cleaned_data/"
    )
)

query.awaitTermination()
```
